```python
import os
import shutil
import glob
```

```python


def get_forcing_dataset(site: str, folder_forcing: str) -> str:
    """
    Search for the first forcing dataset file in the folder that matches the site 
    and does not contain 'IC' in the filename.

    Args:
        site (str): Site identifier string (e.g., 'site=AT-NEU').
        folder_forcing (str): Path to the folder containing forcing dataset files.

    Returns:
        str: Absolute real path of the matching forcing dataset file.

    Raises:
        FileNotFoundError: If no matching file is found.
    """
    # Ensure folder path is absolute
    folder_forcing = os.path.realpath(folder_forcing)
    
    for filename in os.listdir(folder_forcing):
        if site in filename and "IC" not in filename:
            return os.path.realpath(os.path.join(folder_forcing, filename))
    
    raise FileNotFoundError(f"No forcing dataset found for site '{site}' without 'IC' in {folder_forcing}.")
```

```python


def process_filemanager(folder: str):
    """
    Process a fileManager_*.txt file in the given folder:
    - Finds the first fileManager_*.txt whose NAME does NOT contain 'IC'
    - Replaces settingsPath/forcingPath/outputPath with relative './'
    - Removes all '../' from specific lines
    - Saves as fileManager.txt (does not overwrite the original)
    """
    # Step 1: find candidate files (excluding those with 'IC' in filename)
    files = [f for f in glob.glob(os.path.join(folder, "fileManager_*.txt")) 
             if "IC" not in os.path.basename(f)]

    if not files:
        raise FileNotFoundError("No valid fileManager_*.txt found (excluding 'IC' files)")

    target_file = files[0]  # pick the first valid one

    # Step 2: read lines
    with open(target_file, "r") as fh:
        lines = fh.readlines()

    # Step 3: replacements
    replacements = {
        "settingsPath": "settingsPath         './' !  setting_path\n",
        "forcingPath":  "forcingPath          './' !  input_path\n",
        "outputPath":   "outputPath           './' !  output_path\n"
    }

    fixed_lines = []
    for line in lines:
        stripped = line.strip()

        # Replace specific path lines
        if stripped.startswith("settingsPath"):
            fixed_lines.append(replacements["settingsPath"])
        elif stripped.startswith("forcingPath"):
            fixed_lines.append(replacements["forcingPath"])
        elif stripped.startswith("outputPath"):
            fixed_lines.append(replacements["outputPath"])
        # Remove "../" from specific file lines
        elif any(stripped.startswith(prefix) for prefix in [
            "outputControlFile", "decisionsFile", "globalHruParamFile",
            "globalGruParamFile", "vegTableFile", "soilTableFile",
            "generalTableFile", "noahmpTableFile"
        ]):
            fixed_lines.append(line.replace("../", ""))
        else:
            fixed_lines.append(line)

    # Step 4: save to fileManager.txt
    output_path = os.path.join(folder, "fileManager.txt")
    with open(output_path, "w") as fh:
        fh.writelines(fixed_lines)

    print(f"✅ Processed file saved as {output_path}")
```

```python
valid_sites = ["AT-Neu"]  # just one site for now
for site in valid_sites:

    # create a folder for the site
    work_folder_site = os.path.join(work_folder, site)
    summa_folder = os.path.join(work_folder_site, "summa")
    if not os.path.exists(work_folder_site):
        os.makedirs(work_folder_site)
    if not os.path.exists(summa_folder):
        os.makedirs(summa_folder)

    path_forcing = get_forcing_dataset(site, folder_forcing)
    # copy the forcing dataset to the work folder

    shutil.copy(path_forcing, summa_folder)

    # copy all the files from settings folder to the work folder, not any folders just the files to summa folder
    for filename in os.listdir(settings_folder):
        file_path = os.path.join(settings_folder, filename)
        if os.path.isfile(file_path):
            shutil.copy(file_path, summa_folder)

    # inside of settings folder there is a folder with the same name of the site, go inside and copy all the files to the summa folder
    site_folder = os.path.join(settings_folder, site)
    for filename in os.listdir(site_folder):
        file_path = os.path.join(site_folder, filename)
        if os.path.isfile(file_path):
            shutil.copy(file_path, summa_folder)

    process_filemanager(summa_folder)
```
