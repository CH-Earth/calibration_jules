```python
import os

def update_submit_job(folder: str, site: str, exp: str):
    """
    Update submit_job.sh in the given folder:
    - Line 9:   replace job-name with site_exp
    - Line 20:  replace site=... with site=<site>
    - Line 21:  replace needed_metric=... with needed_metric=<exp>
    Saves changes to the same file.
    """
    file_path = os.path.join(folder, "submit_job.sh")

    if not os.path.exists(file_path):
        raise FileNotFoundError(f"{file_path} not found")

    with open(file_path, "r") as f:
        lines = f.readlines()

    # Ensure file has enough lines
    if len(lines) < 21:
        raise ValueError("submit_job.sh does not have enough lines (need at least 21)")

    # Modify line 9 (index 8 in 0-based Python list)
    lines[8] = f"#SBATCH --job-name={site}_{exp}\n"

    # Modify line 20 (index 19)
    lines[19] = f'site="{site}"\n'

    # Modify line 21 (index 20)
    lines[20] = f'needed_metric="{exp}"\n'

    # Write back to the same file
    with open(file_path, "w") as f:
        f.writelines(lines)

    print(f"✅ Updated {file_path} with site={site}, exp={exp}")
```
